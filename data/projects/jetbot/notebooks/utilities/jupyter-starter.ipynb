#### The `[ ]` Status Indicator

**This is text in Markdown format** explaining the bracket indicator to the left of every code cell (it looks like `In [ ]:` in classic Notebook, or just `[ ]:` in JupyterLab). It tells you the run state of that cell:

| Indicator | Meaning |
|-----------|---------|
| `[ ]` (empty) | The cell has **not been run** yet in this session (or the kernel was restarted) |
| `[*]` (asterisk) | The cell is **running right now**, or it is **queued** waiting for the kernel to finish something else |
| `[1]`, `[2]`, `[3]`... (number) | The cell has **finished**. The number is the **order of execution**, not the cell's position in the notebook |

**Things worth knowing:**

- The counter goes up by one every time *any* cell runs. Re-running a cell gives it a new, higher number.
- Because of that, numbers can look out of order (for example `[4]` above `[2]`). That means cells were run out of order, which can cause confusing bugs since variables are shared across the whole session.
- Output for a cell's last expression is labeled with the same number, like `Out[3]:`. Output from `print()` has no `Out` label.
- **A `[*]` that never goes away** usually means the code is stuck in a long loop, waiting for input, or the kernel is busy or hung. Press the stop button in the toolbar (or `I` `I` in command mode) to interrupt. If that fails, restart the kernel with `0` `0`.
- **Kernel > Restart** resets every cell back to `[ ]` and clears all variables, so run your cells from the top afterwards.
- The kernel's overall state is also shown at the top right of the notebook: a filled circle means busy, an empty circle means idle.

````python
# Try it: watch the status indicator to the left of this cell.
# 1. Before you run it, it shows [ ] (empty, never run).
# 2. While it runs, it shows [*] (running). The sleep below keeps it busy for 5 seconds so you can see it.
# 3. When it finishes, it shows a number like [3] (the order this cell was run).
# Run it again and the number will increase, because the counter goes up on every execution.
import time

print("Starting... look for the [*] indicator in 2 seconds")
time.sleep(2)
print("Done! The indicator now shows a number")
````

### Restarting the Kernel and Clearing Outputs

**This is text in Markdown format** explaining two related but *different* actions. The **kernel** is the Python process running behind the notebook. It holds all your variables, imports, and functions in memory. **Outputs** are the results displayed under each cell (printed text, plots, tables) and they are saved inside the `.ipynb` file.

| Action | Where to find it | What it does | What it does NOT do |
|--------|------------------|--------------|---------------------|
| **Restart Kernel** | `Kernel > Restart Kernel...` or `0` `0` in command mode | Wipes all variables, imports, and functions from memory and resets the execution counter | Does not remove outputs already on screen. Does not change your code |
| **Clear All Outputs** | `Edit > Clear Outputs of All Cells` (classic: `Cell > All Output > Clear`) | Removes the displayed results from every cell. Right-click a cell and choose *Clear Outputs* to do one cell | Does not touch memory: your variables still exist. Depending on your Jupyter version, the `[n]` numbers may also reset to `[ ]` |
| **Restart and Clear Outputs** | `Kernel > Restart Kernel and Clear Outputs of All Cells...` | Does both of the above: a completely fresh start with a clean-looking notebook | Does not delete or alter any code |
| **Restart and Run All** | `Kernel > Restart Kernel and Run All Cells...` | Fresh kernel, then runs every cell from top to bottom | Will stop on the first error. Be careful if cells are slow or have side effects |

**When to restart the kernel:**

- The notebook behaves strangely, or you suspect hidden state from running cells out of order
- A cell is stuck on `[*]` and the stop button did not help
- You installed or upgraded a package with `%pip install` and need Python to load the new version
- You want to confirm the notebook really works from top to bottom (use *Restart and Run All*)
- You see "The kernel appears to have died". Jupyter normally restarts it automatically, but your variables are gone

**When to clear outputs:**

- Before sharing or committing the notebook to Git, to keep the file small and avoid leaking data or file paths shown in the results
- When the notebook is cluttered with old results and you want a clean slate

**Good to know:**

- Restarting does **not** delete your notebook or code, only what is in memory. Save your work (`Ctrl+S`) anyway.
- After a restart, widgets from earlier runs stop responding. Re-run the cells that created them.
- Any background work started by the kernel is stopped by a restart.
- Cleared outputs are gone for good once you save, unless you re-run the cells.

````python
# Try it: see what a kernel restart does to memory.
# Step 1: run this cell. It stores a variable in the kernel's memory.
demo_value = 42
print("demo_value is stored in memory:", demo_value)
````

````python
# Step 2: run this cell. It works, because demo_value is still in memory.
# Step 3: now restart the kernel (Kernel > Restart Kernel... or press 0 0 in command mode).
# Step 4: run THIS cell again WITHOUT re-running the cell above.
#         You will get "NameError: name 'demo_value' is not defined", because the restart wiped memory.
# Step 5: run the cell above, then this one again, and it works once more.
# Note: the printed output from earlier cells stays on screen after a plain restart.
# Use "Restart Kernel and Clear Outputs of All Cells..." if you want that wiped too.
print("Reading demo_value:", demo_value)
````

````python
# Variables, loops, and functions work just like in any Python script.
# Variables defined in one cell stay available in later cells (the notebook shares one Python session).

name = "Jupyter"
numbers = [1, 2, 3, 4, 5]

# A for loop
for n in numbers:
    print(f"{name} says: {n} squared is {n**2}")

# A simple function
def average(values):
    # Returns the mean of a list of numbers
    return sum(values) / len(values)

print("Average:", average(numbers))
````

## Widgets

**This is text in Markdown format** introducing widgets. Widgets are interactive controls (sliders, buttons, dropdowns, text boxes) that let you change inputs and see results without editing code.

They come from the `ipywidgets` package. If the next cell fails with `ModuleNotFoundError`, install it first by running `!pip install ipywidgets` in a code cell, then restart the kernel.

````python
# Basic widgets: create a widget object, then display() it to show it in the notebook.
# The widget's current value is always available as widget.value.
import ipywidgets as widgets
from IPython.display import display

slider = widgets.IntSlider(value=5, min=0, max=10, step=1, description="Slider:")
display(slider)

# Drag the slider, then run the next cell to read its current value.
````

````python
# Read the current value of the slider above.
# Widgets hold their state in Python, so you can use it like any variable.
print("The slider is currently set to:", slider.value)
````

````python
# interact(): the quickest way to link a widget to a function.
# Jupyter automatically builds a slider from the range (0, 20) and re-runs the function whenever it moves.
from ipywidgets import interact

def show_square(n=3):
    # This function runs every time the slider changes
    print(f"{n} squared = {n**2}")

interact(show_square, n=(0, 20));
````

````python
# A button with a click handler, plus an Output widget to display results.
# widgets.Output() acts as a container where printed text and plots appear.
button = widgets.Button(description="Click me", button_style="success")
out = widgets.Output()
click_count = 0

def on_click(b):
    # This function runs every time the button is clicked
    global click_count
    click_count += 1
    with out:
        print(f"Button clicked {click_count} time(s)")

button.on_click(on_click)
display(button, out)
````

````python
# Combining several widgets (dropdown + text box) and reacting to changes with observe().
# observe() calls your function whenever the widget's 'value' changes.
dropdown = widgets.Dropdown(options=["Python", "Markdown", "Widgets"], description="Topic:")
textbox = widgets.Text(value="", placeholder="Type something", description="Note:")
result = widgets.Label(value="Pick a topic and type a note")

def update(change):
    result.value = f"You chose {dropdown.value} with note: '{textbox.value}'"

dropdown.observe(update, names="value")
textbox.observe(update, names="value")

display(widgets.VBox([dropdown, textbox, result]))
````
