# Stand-alone test for camera operation

```python
# 1. Imports first
import traitlets
import ipywidgets.widgets as widgets
from jetbot import Camera, bgr8_to_jpeg

# 2. Create the camera
camera = Camera.instance()

# 3. Create the display widget
image = widgets.Image(format='jpeg', width=300, height=300)
display(image)

# 4. Now link them — both camera and image exist at this point
camera_link = traitlets.dlink((camera, 'value'), (image, 'value'), transform=bgr8_to_jpeg)
```

# Always turn off camera when done.

```python
camera.stop()
```
